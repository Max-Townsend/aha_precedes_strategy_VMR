```python
                      
CGDat = allCGData[(allCGData["targetCount"] == 8) &
                  (allCGData["hasAutocorrection"] == 1)]
CGDat['targetPosition'] = CGDat['targetPosition'].apply(lambda x: round(x, 1) if not np.isnan(x) else x)
print("Number of missing targetPositions before filling:", CGDat['targetPosition'].isnull().sum())

                    
CGDat = CGDat.sort_values(['participantNum', 'trialNum'])

                                                                    
cycle_counts = CGDat.groupby(['participantNum', 'cycle']).size()
if not (cycle_counts == 8).all():
    raise ValueError("Some cycles do not contain exactly 8 trials")

                                                     
participant_full_sets = {}
for participant, group in CGDat.groupby('participantNum'):
    unique_pos = group['targetPosition'].dropna().unique()
    if len(unique_pos) != 8:
        raise ValueError(f"Participant {participant} has {len(unique_pos)} unique positions, expected 8")
    participant_full_sets[participant] = set(unique_pos)

                              
for (participant, cycle), cycle_group in CGDat.groupby(['participantNum', 'cycle']):
    positions = cycle_group['targetPosition']
    n_nan = positions.isnull().sum()
    if n_nan > 1:
        raise ValueError(f"More than one NaN in cycle {cycle} for participant {participant}")
    
    non_nan = positions.dropna()
    unique_non_nan = set(non_nan)
    
    full_set = participant_full_sets[participant]
    missing = full_set - unique_non_nan
    if len(missing) > 1:
                                                   
        continue
                                                                                                                                    
    nan_mask = cycle_group['targetPosition'].isnull()
    if sum(nan_mask) > 1:
        print('>1 nan in cycle')
    if len(missing) == 1 and sum(nan_mask) ==1 :
        missing_val = list(missing)[0]
    
        nan_index = cycle_group.index[nan_mask][0]
        
              
        CGDat.at[nan_index, 'targetPosition'] = missing_val

print("Number of missing targetPositions after filling:", CGDat['targetPosition'].isnull().sum())
```

```python

from scipy.stats import pearsonr
importlib.reload(NonGenSSM)
importlib.reload(QLearning)
importlib.reload(HMM)

def get_template_df(CGDat, pp_num, rot_val):
    template = CGDat[(CGDat['participantNum'] == pp_num) &
                     (CGDat['blockNum'] == 0) &
                     (CGDat['blockRot'] == rot_val)].copy()

    template['aim'] = np.nan
    required_cols = ['aim', 'rotation', 'targetPosition', 'participantNum', 'blockNum', 'phase','blockRot']
    for col in required_cols:
        if col not in template.columns:
            template[col] = np.nan 
    return template

def sim_from_ssm(model, pp_idx, template_df):
    params = model.xs[pp_idx]                      
    noise = params[2]
    m_states = np.array(model.mStates[pp_idx])                          
    sim_aims = m_states + np.random.normal(0, noise, len(m_states))
    sim_aims = ((sim_aims + 180) % 360 - 180)
    sim_df = template_df.copy()
    sim_df['aim'] = sim_aims
    sim_df['participantNum'] = 999999                       
    return sim_df, params  

action_centers = np.arange(-180, 181)                  
def sim_from_qlearn(model, pp_idx, template_df):
    params = model.xs[pp_idx]                                 
    noise = params[2]
    policies = model.model_predictive_policies[pp_idx]                                  
    sim_aims = []
    for policy in policies:
        act_idx = np.random.choice(len(action_centers), p=policy)
        act = action_centers[act_idx]
        aim = act + np.random.normal(0, noise)
        sim_aims.append(aim)
    sim_aims = np.array(sim_aims)
    sim_aims = ((sim_aims + 180) % 360 - 180)
    sim_df = template_df.copy()
    sim_df['aim'] = sim_aims
    sim_df['participantNum'] = 999999        
    return sim_df, params

def sim_from_hmm(model, pp_idx, template_df):
    params = model.xs[pp_idx]                                                       
    noise = params[0]
    pi_preds = model.pi_preds[pp_idx]                                         
    policies = model.model_predictive_policies[pp_idx]                                                
    sim_aims = []
    for pi_pred, policy in zip(pi_preds, policies):
        state = np.random.choice([0, 1], p=pi_pred)
        if state == 0:
            aim = np.random.normal(0, noise)
        else:
            act_idx = np.random.choice(len(action_centers), p=policy)
            act = action_centers[act_idx]
            aim = act + np.random.normal(0, noise)
        sim_aims.append(aim)
    sim_aims = np.array(sim_aims)
    sim_aims = ((sim_aims + 180) % 360 - 180)
    sim_df = template_df.copy()
    sim_df['aim'] = sim_aims
    sim_df['participantNum'] = 999999        
    return sim_df, params

if __name__ == '__main__':
    n_starts = 250
    ssms = np.load('ssmsVanilla8AllPPs.npy', allow_pickle=True)
    qlearns = np.load('QLearningVanilla8AllPPs.npy', allow_pickle=True)
    hmms = np.load('HMMVanilla8AllPPs.npy', allow_pickle=True)
    CGDat = allCGData[(allCGData["targetCount"] == 8) &
            (allCGData["hasAutocorrection"] == 0)]
    rots = [-90, -60, -45, -30, -15]
    n_sims_per_pp = 1 
    models = ['SSM', 'QLearn', 'HMM']
    num_models = len(models)
    confusion_matrices = {rot: np.zeros((num_models, num_models)) for rot in rots}                   
    model_to_idx = {name: idx for idx, name in enumerate(models)}
                                                                                       
    param_recovery = {rot: {model: {'true': [], 'fitted': []} for model in models} for rot in rots}

    sim_dfs = []
    metadata = []                                                                 
    dummy_pp_start = 1000000
    for group_idx, rot in enumerate(rots):
        ssm_group = ssms[group_idx]
        qlearn_group = qlearns[group_idx]
        hmm_group = hmms[group_idx]

        num_pps = len(ssm_group.participantNums)
        for pp_idx in range(num_pps):
            pp_num = ssm_group.participantNums[pp_idx]
            template_df = get_template_df(CGDat, pp_num, rot)
         
            for true_model_idx, true_model_name in enumerate(models):
                if true_model_name == 'SSM':
                    true_model = ssm_group
                    sim_func = sim_from_ssm
                elif true_model_name == 'QLearn':
                    true_model = qlearn_group
                    sim_func = sim_from_qlearn
                else:      
                    true_model = hmm_group
                    sim_func = sim_from_hmm
             
                for sim_num in range(n_sims_per_pp):
                    sim_df, true_params = sim_func(true_model, pp_idx, template_df)
                    dummy_pp = dummy_pp_start + len(sim_dfs)
                    sim_df['participantNum'] = dummy_pp
                    sim_dfs.append(sim_df)
                    metadata.append((rot, true_model_name, pp_num, sim_num, true_params))

    big_df = pd.concat(sim_dfs, ignore_index=True)
    
    print(f"Starting model/parameter recovery with {len(metadata)} simulations...")
    start_time = time.time()

    ssm_fit = NonGenSSM.FitShell(df=big_df, condition='none', conVal='none', fitPhase=None, imp=False, n_starts=n_starts)
    ssm_fit.fitRot()

    qlearn_fit = QLearning.FitShell(df=big_df, condition='none', conVal='none', fitPhase=None, n_starts=n_starts)
    qlearn_fit.fitRot()
    
    hmm_fit = HMM.FitShell(df=big_df, condition='none', conVal='none', fitPhase=None, annealing_mode=0, n_starts=n_starts)
    hmm_fit.fitRot()
    
    for i in range(len(metadata)):
        ssm_bic = ssm_fit.bics[i] if len(ssm_fit.bics) > i else np.inf
        qlearn_bic = qlearn_fit.bics[i] if len(qlearn_fit.bics) > i else np.inf
        hmm_bic = hmm_fit.bics[i] if len(hmm_fit.bics) > i else np.inf
        
        bics = [ssm_bic, qlearn_bic, hmm_bic]
        recovered_idx = np.argmin(bics)
        
        rot, true_model_name, _, _, true_params = metadata[i]
        true_idx = model_to_idx[true_model_name]
        confusion_matrices[rot][true_idx, recovered_idx] += 1
        
        if true_model_name == 'SSM':
            fitted_params = ssm_fit.xs[i]
        elif true_model_name == 'QLearn':
            fitted_params = qlearn_fit.xs[i]
        else:
            fitted_params = hmm_fit.xs[i]
        param_recovery[rot][true_model_name]['true'].append(true_params)
        param_recovery[rot][true_model_name]['fitted'].append(fitted_params)
    
    elapsed = time.time() - start_time
    print(f"Completed in {elapsed / 60:.1f} minutes.")
    
    for rot in rots:
        counts = confusion_matrices[rot]
        total_per_true = np.sum(counts, axis=1, keepdims=True)
        proportions = counts / total_per_true                
        print(f"Confusion Matrix for Group {rot} (proportions):")
        df = pd.DataFrame(proportions, index=models, columns=models)
        print(df)
        df.to_csv(f'confusion_{rot}.csv')
    
```

```python
              
y = 'aim'
print('Baseline ttests vs 0')
                                                    
start=[-15,-5,-15,-5]
end=0           
condition = ['a','b']
rots = sorted(exp1["blockRot"].unique())
for i in range(4):
    xs = ['blockTrial','cycle','blockTrial','cycle']
    print(f"EXPERIMENT {1+i//2}{condition[i%2]}")
    for rot in rots:
        stats = DM.tTestPhase(dfs[i],start[i],end,x=xs[i],y=y,dimensions=1,rots=[rot],oneSampleMean=0)[0]
        stats["bonferroni-p"] = min(1, stats["p-val"].values[0] * 5)
        print("Rotation :",rot)
        print(stats)
    print()
```

```python
                                                                                                                                                          

                              
results = []
blockRotLabels = []
targetCount = 1
hasAutocorrection = 0
exp1 = CGDat[(CGDat["targetCount"] == targetCount) &
        (CGDat["hasAutocorrection"] == hasAutocorrection)]

for blockRotValue in sorted(exp1["blockRot"].unique()):
    filtered = exp1[
        (exp1["blockRot"] == blockRotValue) &
        (exp1["phase"] == 'baseline') 
    ]
    participantMeans = (
        filtered
        .groupby("participant")["aim"]
        .mean()
    )
    if len(participantMeans) > 1:
        result = ttest(participantMeans, 0)
        result["bonferroni-p"] = min(1, result["p-val"].values[0] * 5)
        results.append(result)
        blockRotLabels.append(blockRotValue)

print("Experiment 1a baseline ttests")
for blockRot, result in zip(blockRotLabels,results):
    print(f"Rotation {blockRot}: \n{result}")
    print()
```

```python
def bootstrap_anova_ci(data, dv='aim', between=['blockRot', 'learningPhase', 'experiment'],
                       n_boot=1000, ci=95, seed=42):
    np.random.seed(seed)
    aov = pg.anova(data=data, dv=dv, between=between, detailed=True)
    sources = aov['Source'].tolist()                            
    
    boot_np2 = {source: [] for source in sources}
    
    for _ in range(n_boot):
        boot_data = data.sample(n=len(data), replace=True)
        boot_aov = pg.anova(data=boot_data, dv=dv, between=between, detailed=True)
        for source in sources:
            np2_val = boot_aov.loc[boot_aov['Source'] == source, 'np2']
            boot_np2[source].append(np2_val.iloc[0] if len(np2_val) > 0 else np.nan)
    
    lower_p = (100 - ci) / 2
    upper_p = 100 - lower_p
    ci_results = {}
    for source in sources:
        cis = np.percentile(boot_np2[source], [lower_p, upper_p])
        ci_results[source] = (cis[0], cis[1])
    
                                           
    aov['np2_CI'] = [f"[{ci_results[row['Source']][0]:.3f}, {ci_results[row['Source']][1]:.3f}]"
                     for _, row in aov.iterrows()]
    return aov

        
anova_with_ci = bootstrap_anova_ci(participantAverages)
print(anova_with_ci.round(3))
```

```python
descriptives = participantAverages.groupby(['blockRot', 'learningPhase', 'experiment'])['aim'].agg(
    mean='mean',
    std='std',
    n='count'
).reset_index()

                       
confidence = 0.95
t_val = scipy.stats.t.ppf((1 + confidence) / 2, descriptives['n'] - 1)
descriptives['se'] = descriptives['std'] / np.sqrt(descriptives['n'])
descriptives['ci_lower'] = descriptives['mean'] - t_val * descriptives['se']
descriptives['ci_upper'] = descriptives['mean'] + t_val * descriptives['se']

print(descriptives.round(3))
```

```python
import pandas as pd
import numpy as np
import pingouin as pg


cgbt_filtered = cgbt[
    cgbt['aim'].notna() &
    cgbt['blockRot'].notna() &
    cgbt['experiment'].notna() &
    cgbt['participantNum'].notna()
].reset_index(drop=True)

                                                                               
                                                                 
participantAverages = cgbt_filtered.groupby(
    ['participantNum', 'blockRot', 'experiment'],
    as_index=False
)['aim'].mean()

                                                                             
participantAverages = participantAverages.rename(columns={'aim': 'aim_mean'})
participantAverages['subject_str'] = participantAverages['participantNum'].astype(str)

                                               
nParticipants = participantAverages['participantNum'].nunique()
print(f"Number of participants with data after filtering: {nParticipants}")


blockRots = sorted(participantAverages['blockRot'].unique())
experiments = sorted(participantAverages['experiment'].unique())

num_comparisons = len(blockRots)                     

print(f"\nPairwise t-tests: {experiments[0]} vs {experiments[1]} within each blockRot ")

for rot in blockRots:
    data_rot = participantAverages[participantAverages['blockRot'] == rot].copy()
    
    if len(data_rot['experiment'].unique()) != 2:
        print(f"Skipping blockRot {rot}: not exactly 2 experiment levels")
        continue
    
    g1 = data_rot[data_rot['experiment'] == experiments[0]]['aim_mean']
    g2 = data_rot[data_rot['experiment'] == experiments[1]]['aim_mean']
    
                     
    ttest = pg.ttest(g1, g2, alternative='two-sided')
    
    t = ttest['T'].iloc[0]
    p_unc = ttest['p-val'].iloc[0]
    dof = ttest['dof'].iloc[0]
    d = ttest['cohen-d'].iloc[0]
    ci_low, ci_high = ttest['CI95%'].iloc[0] 
    
    p_bonf = min(1, p_unc * num_comparisons)
    
    print(f"\nblockRot {rot}:")
    print(f"  T({dof:.1f}) = {t:.3f}, p_unc = {p_unc:.4f}, p_bonf = {p_bonf:.4f}, d = {d:.3f}")
    print(f"  95% CI diff ({experiments[0]} − {experiments[1]}): [{ci_low:.3f}, {ci_high:.3f}]")
    
                        
    desc = data_rot.groupby('experiment')['aim_mean'].agg(['mean', 'std', 'count']).round(3)
    print("  Descriptives:")
    print(desc)
```
